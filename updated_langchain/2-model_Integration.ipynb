### Google Gemini Chat Model

```python
from langchain.chat_models import init_chat_model
model = init_chat_model(
    "gemini-3.1-flash-lite",
    model_provider = "google_genai"
)
response = model.invoke("can u help me with something?")
response.content[0]["text"]
```

```python
response = model.invoke("write me a 200 words paragraph on artificial intelligence")
response.content[0]["text"]
```

```python
for chunk in model.stream("why do parrots have beautiful feathers?"):
    print(chunk.text, end = "", flush = True)
```

```python
import time

for chunk in model.stream("Tell me a long story about a dragon."):
    print(chunk.text, end="", flush=True)
    time.sleep(0.2)
```
