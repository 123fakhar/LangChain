### Messages and its Types

```python
import os
from dotenv import load_dotenv
load_dotenv()
os.environ["GROQ_API_KEY"]= os.getenv("GROQ_API_KEY")
from langchain.chat_models import init_chat_model
model = init_chat_model(
    "llama-3.3-70b-versatile",
    model_provider = "groq"
)
model
```

```python
model.invoke("who is your daddy meaning in today's context")
```

### List of Messages

```python
from langchain.messages import SystemMessage, AIMessage, HumanMessage
messages = [
    SystemMessage("You are a poetry model"),
    HumanMessage("write a emotional poetry on Imran khan")
]
response = model.invoke(messages)
response.content
```

```python
new_messages = [
    SystemMessage("You are a coding assistant who writes codes"),
    HumanMessage("write me a single python code that involves 2 nested for loops")
]

response = model.invoke(new_messages)
print(response.content)
```

### Detailed System Message

```python
messages_2 = [
    SystemMessage("You are a senior Python Developer with experience in web framework. always provide code examples and explain your reasoning be concise but thorough in your explanations"),
    HumanMessage("How do i create a rest api")
]
response = model.invoke(messages_2)
print(response.content)
```

```python
human_message = HumanMessage(
    content = "Hello I am dying",
    name = "Fakhar Alamgir",
    age = 23
)

response = model.invoke([
    human_message
])

print(response.content)
```
